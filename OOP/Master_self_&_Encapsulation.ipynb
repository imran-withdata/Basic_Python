# **2. Concept of self in oop**

```python
class AtmMachine:
  # constructor
  def __init__(self): # Self is a Property of a calss. AtmMachine is a class and self is it's property. Self always located a Class.
    print(id(self))
    self.pin = ""
    self.balance = 0  # self.pin and self.balance are tow data property of AtmMachine class
    # self.menu()

  def menu(self):
    user_input = input(
        """
        Hello, how would you like to proceed?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """
    )
    if user_input == "1":
      # create a pin
      self.create_pin()

    elif user_input == "2":
        # Change pin
        self.change_pin()

    elif user_input == "3":
      # check balance
      self.check_balance()

    elif user_input == "4":
      # withdraw
      self.withdraw()
    else:
      exit()


  def create_pin(self):
    user_pin =  input("Enter your pin:")
    self.pin = user_pin

    user_balance = int(input("Enter your balance:"))
    self.balance = user_balance

    print("Pin created successfully")
    self.menu()




  def change_pin(self):
    old_pin = input("Enter your old pin:")

    if old_pin == self.pin :
      new_pin = input("Enter your new pin:")
      self.pin = new_pin
      print("Pin changed sucessfully")
      self.menu()
    else:
      print("Invalid pin")
      self.menu()



  def check_balance(self):
    user_pin = input("Enter your pin:")
    if user_pin == self.pin:
      print(f"Your balance id {self.balance}")
    else :
      print("Your pin is incorrect, Pleae try again!")
    self.menu()


  def withdraw(self):
    user_pin = input("Enter your pin:")

    if user_pin == self.pin:
      amount = int(input("Enter the amount: "))

      if amount <= self.balance:
        self.balance = self.balance - amount
        print(f"You have withdraw {amount}. Your new balance is {self.balance}")

      else:
        print("Insufficient balance")

    else:
      print("Invalid your pin, Please try again")

    self.menu()
```

```python
obj = AtmMachine()
```

```python
print(id(obj))
```

```python
class BankAccount:
  def __init__(self, owner, balance):
    self.owner = owner          # Public variable (যেকোনো জায়গা থেকে অ্যাক্সেস করা যাবে)
    self.__balance = balance    # Private variable (ডাবল আন্ডারস্কোর দিয়ে ডাটা হাইড করা হয়েছে)

  # ১. Getter Method: প্রাইভেট ব্যালেন্স দেখার জন্য
  def get_balance(self):
    return self.__balance

  # ২. Setter Method: নিয়মের মাধ্যমে প্রাইভেট ব্যালেন্স পরিবর্তন করার জন্য
  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
      print(f"{amount} taka deposited successfully.")
    else:
      print("Invalid amount!")

# অবজেক্ট তৈরি করা যাক
ac = BankAccount("Buppy", 5000)

# সরাসরি প্রাইভেট ডাটা অ্যাক্সেস করার চেষ্টা (এটি error দেখাবে বা এক্সেস হবে না)
# print(ac.__balance)

# কিন্তু গেটার ও সেটার মেথড দিয়ে এটি নিরাপদে করা সম্ভব
print("Owner:", ac.owner)
print("Initial Balance:", ac.get_balance())
ac.deposit(1500)
print("Updated Balance:", ac.get_balance())
```

```python
Encapsulation (এনক্যাপসুলেশন) হলো অবজেক্ট-ওরিয়েন্টেড প্রোগ্রামিংয়ের (OOP) অন্যতম একটি গুরুত্বপূর্ণ ধারণা।

নিচে বাংলায় এর সহজ ব্যাখ্যা দেওয়া হলো এবং আপনার সিলেক্ট করা সেলে (L-1v0KGCu1ez) একটি কোড লিখে দেওয়া হলো:

১. Encapsulation কী?
En-capsulation শব্দটির উৎপত্তি হয়েছে Capsule থেকে। একটি ক্যাপসুলের ভেতর যেমন ওষুধ সুরক্ষিত বা বন্দি থাকে, ঠিক তেমনি একটি ক্লাসের ভেতরে তার ডাটা (variables) এবং মেথড (functions)-কে একসাথে আটকে রাখাকেই এনক্যাপসুলেশন বলে।
এর প্রধান লক্ষ্য হলো ক্লাসের ভেতরের সংবেদনশীল ডাটা যেন বাইরের কেউ সরাসরি অ্যাক্সেস বা পরিবর্তন করতে না পারে।
২. এটি কীভাবে কাজ করে?
পাইথনে ডাটা হাইড বা প্রাইভেট করার জন্য ভ্যারিয়েবলের নামের শুরুতে দুটি আন্ডারস্কোর __ (double underscore) ব্যবহার করা হয়।

যেমন: self.__balance একটি প্রাইভেট ভ্যারিয়েবল, যা ক্লাসের বাইরে থেকে সরাসরি দেখা বা পরিবর্তন করা যায় না।
এই প্রাইভেট ডাটা দেখতে বা পরিবর্তন করতে আমরা বিশেষ মেথড ব্যবহার করি, যেগুলোকে Getter এবং Setter মেথড বলা হয়।
```
